```python

# EXPERIMENT 2 - ACTIVATION FUNCTIONS

# Roll No      : 24BAD016


# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Generate Input Values
x = np.linspace(-10, 10, 500)


# Activation Functions


# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh Function
def tanh(x):
    return np.tanh(x)

# ReLU Function
def relu(x):
    return np.maximum(0, x)

# Compute Outputs
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)


# Plot the Functions

plt.figure(figsize=(12,6))

plt.plot(x, y_sigmoid, label="Sigmoid", linewidth=2)
plt.plot(x, y_tanh, label="Tanh", linewidth=2)
plt.plot(x, y_relu, label="ReLU", linewidth=2)

plt.title("Comparison of Activation Functions", fontsize=16)
plt.xlabel("Input (x)", fontsize=12)
plt.ylabel("Output", fontsize=12)
plt.grid(True)
plt.legend()
plt.show()


# Individual Graphs


plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
plt.plot(x, y_sigmoid, color='blue')
plt.title("Sigmoid")
plt.grid(True)

plt.subplot(1,3,2)
plt.plot(x, y_tanh, color='green')
plt.title("Tanh")
plt.grid(True)

plt.subplot(1,3,3)
plt.plot(x, y_relu, color='red')
plt.title("ReLU")
plt.grid(True)

plt.tight_layout()
plt.show()

# Comparison Table


print("ACTIVATION FUNCTION COMPARISON")
print("-----------------------------------------------")
print("1. Sigmoid")
print("   Output Range       : (0, 1)")
print("   Saturation Region  : Very high and very low inputs")
print("   Gradient           : Vanishing gradient problem")
print("   Efficiency         : Slow due to exponential")
print("   Applications       : Binary classification output layer\n")

print("2. Tanh")
print("   Output Range       : (-1, 1)")
print("   Saturation Region  : Large positive and negative inputs")
print("   Gradient           : Less vanishing than sigmoid")
print("   Efficiency         : Moderate")
print("   Applications       : Hidden layers in RNNs\n")

print("3. ReLU")
print("   Output Range       : [0, ∞)")
print("   Saturation Region  : Negative inputs become zero")
print("   Gradient           : No vanishing for positive inputs")
print("   Efficiency         : Very fast")
print("   Applications       : Hidden layers in deep neural networks")
```
